### Thermo-Mechanical Verification in Axisymmetry (R–Z) — Uniform Heating of an Annulus

Axisymmetric thermo-elastic verification on an annular domain uniformly heated by ΔT (small-strain, linear, isotropic: E, ν, α). Inner/outer radii are traction-free. Validation focuses on displacements: radial $u_r$ and $u_z$ sampled along the mid-height line $z=z_{\text{mid}}$ and the mid-radius $r=(r_i+r_e)/2$.

```python
from trustutils import run
run.reset()
run.initBuildDirectory()
```

```python
E = 160e9
nu = 0.31
alpha = 15e-6
T0 = 20
T = 500
run.addCaseFromTemplate("jdd.data", "results", {"E" : E, "nu" : nu, "alpha": alpha, "T0": T0, "T": T})
run.runCases()
run.tablePerf()
```

```python
from trustutils import plot
import numpy as np

fig = plot.Graph()
fig.addResidu(f"{run.BUILD_DIRECTORY}/results/jdd.dt_ev", label=" ")
fig.scale("linear", "log")

r_interval = np.linspace(0, 1, 100)
z_interval = np.linspace(0, 2, 100)

# deplacement
fig = plot.Graph("ur en fonction de r a z=0.5")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd_DX.son", marker="x", label=f"TRUST", compo=0)
sol = [alpha * (T - T0) * r for r in r_interval]
fig.add(r_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("uz en fonction de r a z=0.5")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd_DX.son", marker="x", label=f"TRUST", compo=1)
sol = [alpha * (T - T0) * 0.5 for r in r_interval]
fig.add(r_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("ur en fonction de z a r=0.125")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd_DY.son", marker="x", label=f"TRUST", compo=0)
sol = [alpha * (T - T0) * 0.5 for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("uz en fonction de z a r=0.241")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd_DY.son", marker="x", label=f"TRUST", compo=1)
sol = [alpha * (T - T0) * z for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")

fig = plot.Graph(nY=2)
fig.addPlot(0, title="u_r(r)")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd_DX.son", marker="x", label=f"TRUST", compo=0)
sol = [alpha * (T - T0) * r for r in r_interval]
fig.add(r_interval, sol, label=f"Exact solution", marker="--")
fig.label("r [m]", "Displacement [m]")
fig.addPlot(1, title="u_z(z)")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd_DY.son", marker="x", label=f"TRUST", compo=1)
sol = [alpha * (T - T0) * z for z in z_interval]
fig.add(z_interval, sol, label=f"Exact solution", marker="--")
fig.label("z [m]", "Displacement [m]")
```
