### Static Linear Elasticity in Axisymmetric (R–Z) — Analytic Verification

This notebook runs a compact suite of **static linear-elastic** problems in axisymmetry (R–Z).  
Material parameters are **E = 160 GPa**, **ν = 0.31**; Lamé **μ, λ** are computed from (E, ν).  
Each case is solved and **checked against a closed-form analytic solution**. Axisymmetric operators (including **1/r** terms) and **axis regularity at r = 0** are handled explicitly.

**Test cases**
- **jdd0** — annulus; Dirichlet displacements on all boundaries + body force.
- **jdd1** — annulus; same physics as *jdd0* (variant setup/mesh) + body force.
- **jdd2** — annulus; mixed BCs: Dirichlet + prescribed traction.
- **jdd3** — annulus; **frictionless (roller) boundary** where stated (**uₙ = 0**, tangential free) + prescribed traction.
- **jdd4** — solid cylinder including the axis; **axis regularity at r = 0** (**uᵣ = 0**) + prescribed traction on outer boundaries (roller where stated).

```python
from trustutils import run
run.reset()
run.initBuildDirectory()
```

```python
h = 0.1
z0 = 0
ri = 0.24
t_w = 2e-3
re = ri + t_w

E = 160e9
nu = 0.31
# E = 29.9
# nu = 0.495
mu = E / (2 * (1 + nu))
lambda_ = E * nu / ((1 + nu) * (1 - 2 *nu))
alpha = 1e-1
beta = 1e-1
run.addCaseFromTemplate("jdd0.data", "results", {"E" : E, "nu" : nu, "t_w": t_w, "ri": ri, "re": re, "mu": mu, "lambda": lambda_, "h": h, "z0": z0})
run.addCaseFromTemplate("jdd1.data", "results", {"E" : E, "nu" : nu, "t_w": t_w, "ri": ri, "re": re, "mu": mu, "lambda": lambda_, "h": h, "z0": z0})
run.addCaseFromTemplate("jdd2.data", "results", {"E" : E, "nu" : nu, "t_w": t_w, "ri": ri, "re": re, "mu": mu, "lambda": lambda_, "h": h, "z0": z0})
run.addCaseFromTemplate("jdd3.data", "results", {"E" : E, "nu" : nu, "t_w": t_w, "ri": ri, "re": re, "mu": mu, "lambda": lambda_, "h": h, "z0": z0, "alpha": alpha, "beta": beta})
run.addCaseFromTemplate("jdd4.data", "results", {"E" : E, "nu" : nu, "t_w": t_w, "ri": ri, "re": re, "mu": mu, "lambda": lambda_, "h": h, "z0": z0, "alpha": alpha, "beta": beta})
run.runCases()
run.tablePerf()
```

```python
from trustutils import plot
import numpy as np
r_interval = np.linspace(ri, ri + t_w, 100)
z_interval = np.linspace(z0, z0 + h, 100)

fig = plot.Graph()
fig.addResidu(f"{run.BUILD_DIRECTORY}/results/jdd0.dt_ev", label=" ")
fig.scale("linear", "log")

# deplacement
fig = plot.Graph("ur en fonction de r a z=0.05")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_DX.son", marker="x", label=f"TRUST", compo=0)
sol = [r * np.cos(np.pi * 0.05 / h) for r in r_interval]
fig.add(r_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("uz en fonction de r a z=0.05")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_DX.son", marker="x", label=f"TRUST", compo=1)
sol = [-2.0 * np.sin(np.pi * 0.05 / h) * h / np.pi for r in r_interval]
fig.add(r_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("uz en fonction de z a r=0.241")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_DY.son", marker="x", label=f"TRUST", compo=1)
sol = [-2.0 * np.sin(np.pi * z / h) * h / np.pi for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("ur en fonction de z a r=0.241")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_DY.son", marker="x", label=f"TRUST", compo=0)
sol = [0.241 * np.cos(np.pi * z / h) for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")

# deformation
fig = plot.Graph("eps_r en fonction de z a r=0.241")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_EPSILON.son", marker="x", label=f"TRUST", compo=0)
sol = [np.cos(np.pi * z / h) for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("eps_z en fonction de z a r=0.241")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_EPSILON.son", marker="x", label=f"TRUST", compo=1)
sol = [-2.0 * np.cos(np.pi * z / h) for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")

# contraintes
fig = plot.Graph("sigma_r en fonction de z a r=0.241")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_SIGMA.son", marker="x", label=f"TRUST", compo=0)
sol = [2.0 * mu * np.cos(np.pi * z / h) for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("sigma_z en fonction de z a r=0.241")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_SIGMA.son", marker="x", label=f"TRUST", compo=1)
sol = [-4.0 * mu * np.cos(np.pi * z / h) for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")

fig = plot.Graph("sigma_theta en fonction de z a r=0.241")
fig.addSegment(f"{run.BUILD_DIRECTORY}/results/jdd0_SIGMA.son", marker="x", label=f"TRUST", compo=2)
sol = [2.0 * mu * np.cos(np.pi * z / h) for z in z_interval]
fig.add(z_interval, sol, label=f"formule", marker="--")
```
